```python
import torch.nn as nn
import torch

class MySimpleNN(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out

  def loss_function(self,y_pred,y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = y_pred.clamp(min=epsilon, max=1-epsilon)

    # Calculate loss
    loss = - (y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
# Define hyperparameters locally
learning_rate = 0.1
epochs = 25

# Initialize model with correct number of features
model = MySimpleNN(X_train_tensor.shape[1])

# Define loop
for epoch in range(epochs):

  # Forward pass (ensure inputs are float tensors for PyTorch layers)
  y_pred = model(X_train_tensor.float())

  # Compute loss (ensure targets match target shape and are float tensors)
  loss = model.loss_function(y_pred, y_train_tensor.float().unsqueeze(1))

  # Backward pass
  loss.backward()

  # Parameter update
  with torch.no_grad():
    model.linear.weight.data -= learning_rate * model.linear.weight.grad
    model.linear.bias.data -= learning_rate * model.linear.bias.grad

  # Zero gradients
  model.linear.weight.grad.zero_()
  model.linear.bias.grad.zero_()

  # Print loss in each epoch
  print(f'Epoch: {epoch + 1} loss: {loss.item():.4f}')
```

```python
model.linear.weight
```

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor.float())
  y_pred = (y_pred > 0.5).float()

  accuracy = (y_pred == y_test_tensor.float().unsqueeze(1)).float().mean()
  print(f'Accuracy: {accuracy.item():.4f}')

print("Predictions:", y_pred.flatten())
print("Actual Labels:", y_test_tensor)
```
