3. Reduction operation

```python
e=torch.randint(size=(2,3), low=0 , high =10,dtype=torch.float32)
e
```

```python
# mean
torch.mean(e)
# mean along col
torch.mean(e,dim=0)
# mean along row
torch.mean(e,dim=1)
```

```python
# median
torch.median(e)
```

```python
# max and min
torch.max(e)
torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# arg max
# This tells the biggest item's position.
torch.argmax(e)
```

Inplace operations

```python
m = torch.rand(2,3)
n = torch.rand(2,3)
print(m)
print(n)
```

When we do m+n this will store the value by making another tensor and it will consume more memory space if we are working on large datasets.
That's why we use inplace operations to store value in one of the variable created above


```python
m.add_(n)
```
